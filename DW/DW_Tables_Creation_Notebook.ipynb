```python

from pyspark.sql.functions import (
    explode, sequence, to_date, to_timestamp, col,
    dayofweek, weekofyear, month, quarter, year,
    date_format, when, lit
)

# =========================================
# 1. RESET TABLE (IMPORTANT FIX)
# =========================================

spark.sql("DROP TABLE IF EXISTS dim_date")

spark.sql("""
CREATE TABLE dim_date (
    full_date TIMESTAMP,
    day_of_week INT,
    day_name STRING,
    week_number INT,
    month_number INT,
    month_name STRING,
    quarter INT,
    year INT,
    is_weekend BOOLEAN,
    is_holiday BOOLEAN
)
USING DELTA
""")

# =========================================
# 2. GENERATE DATE RANGE
# =========================================

df_dates = spark.sql("""
SELECT explode(
    sequence(
        to_date('2015-01-01'),
        to_date('2030-12-31'),
        interval 1 day
    )
) AS full_date
""")

# =========================================
# 3. TRANSFORM DATA
# =========================================

df_dim_date = df_dates.select(
    to_timestamp(col("full_date")).alias("full_date"),

    dayofweek("full_date").alias("day_of_week"),  # INT (fixed)
    date_format("full_date", "EEEE").alias("day_name"),

    weekofyear("full_date").alias("week_number"),
    month("full_date").alias("month_number"),
    date_format("full_date", "MMMM").alias("month_name"),
    quarter("full_date").alias("quarter"),
    year("full_date").alias("year"),

    when(dayofweek("full_date").isin(1, 7), True).otherwise(False).alias("is_weekend"),
    lit(False).alias("is_holiday")
)

# =========================================
# 4. LOAD DATA (SAFE MODE)
# =========================================

df_dim_date.write \
    .format("delta") \
    .mode("overwrite") \
    .option("overwriteSchema", "true") \
    .saveAsTable("dim_date")
```
