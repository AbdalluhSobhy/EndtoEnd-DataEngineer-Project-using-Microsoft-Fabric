# Create STG Database

```python
spark.sql("SHOW DATABASES").show(truncate=False)
```

```python
spark.sql("USE dbo")
```

```python
# =========================
# Game Tables
# =========================

spark.sql("""
CREATE TABLE IF NOT EXISTS game_titles (
    game_id INT,
    game_name STRING,
    ModifiedDate TIMESTAMP
)
USING DELTA
""")

spark.sql("""
CREATE TABLE IF NOT EXISTS game_genres (
    game_id INT,
    genre STRING,
    ModifiedDate TIMESTAMP
)
USING DELTA
""")

spark.sql("""
CREATE TABLE IF NOT EXISTS game_prices (
    game_id INT,
    price DECIMAL(10,2),
    discount_percentage DECIMAL(5,2),
    tax_percentage DECIMAL(5,2),
    ModifiedDate TIMESTAMP
)
USING DELTA
""")

spark.sql("""
CREATE TABLE IF NOT EXISTS game_metadata (
    game_id INT,
    release_date DATE,
    platform STRING,
    rating DECIMAL(3,1),
    ModifiedDate TIMESTAMP
)
USING DELTA
""")

# =========================
# User Tables
# =========================

spark.sql("""
CREATE TABLE IF NOT EXISTS user_basic (
    user_id INT,
    username STRING,
    status STRING,
    ModifiedDate TIMESTAMP
)
USING DELTA
""")

spark.sql("""
CREATE TABLE IF NOT EXISTS user_contact (
    user_id INT,
    email STRING,
    phone STRING,
    ModifiedDate TIMESTAMP
)
USING DELTA
""")

spark.sql("""
CREATE TABLE IF NOT EXISTS user_location (
    user_id INT,
    city_id INT,
    ModifiedDate TIMESTAMP
)
USING DELTA
""")

spark.sql("""
CREATE TABLE IF NOT EXISTS user_activity (
    user_id INT,
    created_at TIMESTAMP,
    ModifiedDate TIMESTAMP
)
USING DELTA
""")

# =========================
# Orders Table
# =========================

spark.sql("""
CREATE TABLE IF NOT EXISTS orders (
    order_id INT,
    user_id INT,
    game_id INT,
    game_name STRING,
    genre STRING,
    country STRING,
    order_date TIMESTAMP,
    price DECIMAL(10,2),
    quantity INT,
    discount_amount DECIMAL(10,2),
    tax_amount DECIMAL(10,2),
    total_amount DECIMAL(10,2),
    is_refunded BOOLEAN,
    refund_date TIMESTAMP,
    payment_method STRING,
    ModifiedDate TIMESTAMP
)
USING DELTA
""")

# =========================
# Gameplay Tables
# =========================

spark.sql("""
CREATE TABLE IF NOT EXISTS game_sessions (
    session_id INT,
    user_id INT,
    game_id INT,
    hours_played DECIMAL(10,2),
    session_date TIMESTAMP,
    device_type STRING,
    ModifiedDate TIMESTAMP
)
USING DELTA
""")

spark.sql("""
CREATE TABLE IF NOT EXISTS trophies (
    user_id INT,
    game_id INT,
    trophy_name STRING,
    trophy_type STRING,
    earned_date TIMESTAMP,
    ModifiedDate TIMESTAMP
)
USING DELTA
""")

# =========================
# Location Tables
# =========================

spark.sql("""
CREATE TABLE IF NOT EXISTS countries (
    country_id INT,
    country_name STRING,
    region STRING,
    ModifiedDate TIMESTAMP
)
USING DELTA
""")

spark.sql("""
CREATE TABLE IF NOT EXISTS states (
    state_id INT,
    state_name STRING,
    country_id INT,
    ModifiedDate TIMESTAMP
)
USING DELTA
""")

spark.sql("""
CREATE TABLE IF NOT EXISTS cities (
    city_id INT,
    city_name STRING,
    state_id INT,
    ModifiedDate TIMESTAMP
)
USING DELTA
""")
```
